# World Happiness Report 2023 – Exploratory Data Analysis

**Project:** World Happiness EDA  
**Dataset:** World Happiness Report 2023 (`WHR2023.csv`)  
**Tools:** Python, Pandas, NumPy, Matplotlib, Seaborn, Google Colab

This notebook follows the analysis shown in the supplied PDF. It explores happiness scores across 137 countries and studies their relationship with social and economic factors.

## 1. Import libraries and upload dataset

Upload either the original `archive.zip` file (containing `WHR2023.csv`) or the CSV file itself when prompted.

```python
import os
import zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

uploaded = files.upload()
filename = next(iter(uploaded))

if filename.lower().endswith(".zip"):
    with zipfile.ZipFile(filename, "r") as zip_ref:
        zip_ref.extractall("/content/World_Happiness_Project")
    csv_path = "/content/World_Happiness_Project/WHR2023.csv"
elif filename.lower().endswith(".csv"):
    csv_path = filename
else:
    raise ValueError("Please upload archive.zip or WHR2023.csv")

df = pd.read_csv(csv_path)
print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)
display(df.head())
```

## 2. Explore the dataset

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values in Each Column:")
display(df.isnull().sum())
```

```python
print("Total Duplicate Records:")
print(df.duplicated().sum())
```

```python
df.describe()
```

## 3. Data cleaning

```python
# Fill the missing value in Healthy life expectancy with its median
df['Healthy life expectancy'] = df['Healthy life expectancy'].fillna(
    df['Healthy life expectancy'].median()
)

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Total Duplicate Records:")
print(df.duplicated().sum())

df.describe()
```

```python
print("Final Dataset Rows:", df.shape[0])
print("Final Dataset Columns:", df.shape[1])
```

## 4. Top 10 happiest countries

```python
top_10 = df.sort_values(
    by='Ladder score',
    ascending=False
).head(10)

print("Top 10 Happiest Countries")
display(top_10[['Country name', 'Ladder score']])
```

## 5. Top 10 least-happy countries

```python
bottom_10 = df.sort_values(
    by='Ladder score',
    ascending=True
).head(10)

print("Top 10 Least Happy Countries")
display(bottom_10[['Country name', 'Ladder score']])
```

## 6. Happiness score summary

```python
print("Average Happiness Score:",
      df['Ladder score'].mean())

print("Maximum Happiness Score:",
      df['Ladder score'].max())

print("Minimum Happiness Score:",
      df['Ladder score'].min())

print("Median Happiness Score:",
      df['Ladder score'].median())
```

## 7. Correlation analysis

```python
factors = [
    'Logged GDP per capita',
    'Social support',
    'Healthy life expectancy',
    'Freedom to make life choices',
    'Generosity',
    'Perceptions of corruption'
]

correlation = df[factors + ['Ladder score']].corr()

print("Correlation with Happiness Score:")
display(
    correlation['Ladder score']
    .sort_values(ascending=False)
)
```

```python
print("Social Support Analysis")

print("Average Social Support:",
      df['Social support'].mean())

print("Correlation between Social Support and Happiness:",
      df['Social support'].corr(df['Ladder score']))
```

## 8. Visualizations

### 8.1 Top 10 happiest countries – bar chart

```python
plt.figure(figsize=(10, 6))
plt.barh(top_10['Country name'], top_10['Ladder score'])
plt.title('Top 10 Happiest Countries - 2023')
plt.xlabel('Happiness Score')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

### 8.2 Happiness score of top 20 countries – line chart

```python
rank_data = df.sort_values(
    'Ladder score',
    ascending=False
).head(20)

plt.figure(figsize=(10, 6))
plt.plot(
    rank_data['Country name'],
    rank_data['Ladder score'],
    marker='o'
)
plt.title('Happiness Score of Top 20 Countries')
plt.xlabel('Country')
plt.ylabel('Happiness Score')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

### 8.3 Distribution of happiness scores – histogram

```python
plt.figure(figsize=(10, 6))
plt.hist(
    df['Ladder score'],
    bins=15,
    edgecolor='black'
)
plt.title('Distribution of Happiness Scores')
plt.xlabel('Happiness Score')
plt.ylabel('Number of Countries')
plt.show()
```

### 8.4 GDP vs happiness score – scatter plot

```python
plt.figure(figsize=(10, 6))
plt.scatter(
    df['Logged GDP per capita'],
    df['Ladder score']
)
plt.title('GDP vs Happiness Score')
plt.xlabel('Logged GDP per Capita')
plt.ylabel('Happiness Score')
plt.show()
```

### 8.5 Country happiness distribution – pie chart

```python
df['Happiness Category'] = pd.cut(
    df['Ladder score'],
    bins=[0, 4, 5, 6, 7, 10],
    labels=['Below 4', '4-5', '5-6', '6-7', 'Above 7'],
    include_lowest=True,
    right=False
)

df['Happiness Category'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(8, 8)
)
plt.title('Country Happiness Distribution')
plt.ylabel('')
plt.show()
```

### 8.6 Correlation heatmap

```python
plt.figure(figsize=(10, 8))

columns = [
    'Ladder score',
    'Logged GDP per capita',
    'Social support',
    'Healthy life expectancy',
    'Freedom to make life choices',
    'Generosity',
    'Perceptions of corruption'
]

sns.heatmap(
    df[columns].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('World Happiness Correlation Heatmap')
plt.tight_layout()
plt.show()
```
